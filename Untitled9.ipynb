```python
X = df.drop("quality", axis=1)
y = df["quality"]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)
```

```python
pca = PCA() #obj created
X_pca = pca.fit_transform(X_scaled)
print("Original shape:", X.shape)
print("PCA shape:", X_pca.shape)
```

```python
explained_variance = pca.explained_variance_ratio_
print("Explained Variance Ratio:")
for i, variance in enumerate(explained_variance, start=1):
    print(f"PC{i}: {variance:.4f}")
```

```python
components = range(1, len(explained_variance) + 1)
plt.bar(
    components,
    explained_variance
)
plt.xlabel("Principal Component")
plt.ylabel("Explained Variance Ratio")
plt.title("Explained Variance by Principal Component")
plt.xticks(components)
plt.show()
```

```python
cumulative_variance = np.cumsum(explained_variance)
print("Cumulative Explained Variance:")
for i, variance in enumerate(cumulative_variance, start=1):
    print(f"First {i} components: {variance:.4f}")
```

```python
plt.plot(
    components,
    cumulative_variance,
    marker="o"
)
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Cumulative Explained Variance")
plt.xticks(components)
plt.grid()
plt.show()
```

```python
n_components = np.argmax(cumulative_variance >= 0.95) + 1
print("Number of components required for 95% variance:", n_components)
```

```python
n_components_90 = np.argmax(cumulative_variance >= 0.90) + 1
print("Components for 90% variance:", n_components_90)
```

```python
pca_final = PCA(n_components=n_components)
X_reduced = pca_final.fit_transform(X_scaled)
print("Original dimensions:", X.shape)
print("Reduced dimensions:", X_reduced.shape)
```

```python
pca_columns = [
    f"PC{i}"
    for i in range(1, n_components + 1)
]
df_pca = pd.DataFrame(
    X_reduced,
    columns=pca_columns
)
df_pca["quality"] = y.values
print(df_pca.head())
```

```python
scatter = plt.scatter(
    df_pca["PC1"],
    df_pca["PC2"],
    c=df_pca["quality"],
    cmap="viridis",
    alpha=0.7
)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Wine Dataset After PCA")
plt.colorbar(scatter, label="Wine Quality")
plt.show()
```

```python
print("Original number of features:", X.shape[1])
print("Reduced number of features:", X_reduced.shape[1])
print(
    f"Dimensionality reduced from "
    f"{X.shape[1]} to {X_reduced.shape[1]} features."
)
```
